```python
import pandas as pd
import matplotlib.pyplot as plt

labels = ['rule', 'rule_long_term']

data = {
    'rule': {
        'success': [True, False, True, True, False, False, False],
        'num_actions': [5, 34, 13, 6, 5, 35, 35]
    },
    'rule_long_term': {
        'success': [True, False, True, True, True, True, False],
        'num_actions': [5, 35, 11, 5, 28, 24, 35]
    }
}


# Convert to DataFrame
rule_df = pd.DataFrame(data['rule'])
rule_long_term_df = pd.DataFrame(data['rule_long_term'])

games = range(1, len(rule_df) + 1)

# Create two subplots
fig, axes = plt.subplots(2, 1, figsize=(10, 10))

# 1. Plot Success Comparison
axes[0].plot(games, rule_df['success'], 'o-', label='Rule Success', color='green')
axes[0].plot(games, rule_long_term_df['success'], 'o--', label='Rule Long Term Success', color='darkgreen')
axes[0].set_xlabel('Number of Games')
axes[0].set_ylabel('Success (1=True, 0=False)')
axes[0].set_title('Success Comparison: Rule vs Rule Long Term')
axes[0].set_yticks([0, 1])
axes[0].legend()

# 2. Plot Number of Actions Comparison
axes[1].plot(games, rule_df['num_actions'], 's-', label='Rule Actions', color='blue')
axes[1].plot(games, rule_long_term_df['num_actions'], 's--', label='Rule Long Term Actions', color='darkblue')
axes[1].set_xlabel('Number of Games')
axes[1].set_ylabel('Number of Actions')
axes[1].set_title('Number of Actions Comparison: Rule vs Rule Long Term')
axes[1].legend()

# Adjust layout
plt.tight_layout()
plt.show()
```
